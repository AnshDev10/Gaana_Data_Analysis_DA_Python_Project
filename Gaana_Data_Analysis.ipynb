##3. Data Cleaning

```python
#here will convert the duration into seconds for that we will create a new column and split on the basis of ':'
#we will fist consider 'lst = []' that will take the values that are splitted
#now using if-elif-else condition we divide the time into hour,mins,sec and will split them on the basis of ':'

lst = []
for i in df['duration']:
  if len(i.split(':')) == 3:
    hour = int(i.split(':')[0]) * 3600
    min = int(i.split(':')[1]) * 60
    sec = int(i.split(':')[2])
    lst.append(hour+min+sec)

  elif len(i.split(':')) == 2:
    min = int(i.split(':')[0]) * 60
    sec = int(i.split(':')[1])
    lst.append(min+sec)

  else:
    lst.append(sec)

df['Duration_in_sec'] = lst
```

```python
df.drop('duration',axis=1,inplace=True)
```

###4.10) Average Duration of Songs for each language

```python
df.groupby('language')['Duration_in_sec'].mean().sort_values(ascending=False)
```

```python
avg_duration_by_language = df.groupby('language')['Duration_in_sec'].mean().sort_values(ascending=False)
plt.figure(figsize=(14, 7))
sns.barplot(x=avg_duration_by_language.index, y=avg_duration_by_language.values, palette='crest')
plt.title('Average Song Duration per Language (in seconds)')
plt.xlabel('Language')
plt.ylabel('Average Duration (seconds)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Insight: Average Duration of Songs by Language

**Urdu** and **Punjabi** songs tend to have the longest average durations, both exceeding 400 seconds (approximately 6-7 minutes). In contrast, languages like Haryanvi have a shorter average duration. This might reflect cultural differences in song structure, genre prevalence, or storytelling traditions within music across different languages.
